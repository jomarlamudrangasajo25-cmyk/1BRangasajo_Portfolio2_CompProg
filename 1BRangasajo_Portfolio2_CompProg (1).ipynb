pH Level Classifier

During chemical reactions, especially those where neutralization
is involved, pH often change when ions react. Because these
changes determine the final state of the solution, an automatic
classifier is useful to instantly identify whether 
a reaction has reached a neutral, acidic, or basic endpoint.

```python
x= float(input('The pH of the solution is: '))
if x > 7:
    print('The solution is basic')
elif x == 7:
    print('The solution is neutral')
else: 
    print('The solution is acidic')
```
